```python
import pandas as pd
with open("Data1.txt", 'r') as f:
  lines = f.readlines()
df = pd.DataFrame(lines, columns=['text_content'])
display(df.head())
```

```python
df.shape
```

```python
df['text_content'] = df['text_content'].str.lower()
```

```python
df
```

Removing HTML Tags

```python
import re
def remove_html_tags(text):
  pattern = re.compile('<.*?>')
  return pattern.sub(r'', text)
```

```python
df['text_content'] = df['text_content'].apply(remove_html_tags)
```

```python
df['text_content'][15]
```

Removing link text

```python
def remove_url(text):
  pattern = re.compile(r'https?://\S+|www\.\S+')
  return pattern.sub(r'', text)
```

```python
df['text_content'] = df['text_content'].apply(remove_url)
df['text_content']
```

Removing Bracketed Numbers/Citations

```python
df['text_content'] = df['text_content'].str.replace(r'\[\d+\]', '', regex=True)
```

Removing \n

```python
df['text_content'] = df['text_content'].str.replace('\n','')
```

Removing punctuation

```python
import string,time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punc(text):
  for char in exclude:
    text = text.replace(char,'')
  return text
```

```python
df['text_content'] = df['text_content'].apply(remove_punc)
```

Removing Inflection(Lemmatization Method)

```python
from nltk.stem import WordNetLemmatizer
wnl= WordNetLemmatizer()
```

```python
nltk.download('wordnet')
```

```python
def lemmatize_text(text):
  words = text.split()
  lemmatized_words = [wnl.lemmatize(word) for word in words]
  return ' '.join(lemmatized_words)

lemmatize_df2 = lemmatize_text(df2)
print(lemmatize_df2)
```
